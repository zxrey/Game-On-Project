```python
import numpy as np
import pandas as pd
import matplotlib
import re
```

```python
file = "/mnt/c/Users/Usuario/Downloads/steam_games.csv"
df = pd.read_csv(file)
```

```python
#Limpieza de 'recent_reviews'
def extract_percentage(text):
    if not isinstance(text, str):
        return None  # or np.nan, or 0
    match = re.search(r'([-]?\d+)%', text)
    if match:
        return int(match.group(1))
    return None

df['review_percentage'] = df['recent_reviews'].apply(extract_percentage)
```

```python
# Se eliminan columnas que no son relevantes para el análisis:
df = df.drop(columns=['types', 'all_reviews','desc_snippet', 'recent_reviews', 'developer', 'publisher', 'achievements', 'mature_content', 'minimum_requirements', 'recommended_requirements', 'discount_price'])
```

```python
# limpieza de 'genre'
df['genre'] = df['genre'].str.replace(',', ', ', regex=False)
```

```python
# limpieza de 'popular_tags'
df['popular_tags'] = df['popular_tags'].str.replace(',', ', ', regex=False)
```

```python
# limpieza de 'game_details'
df['game_details'] = df['game_details'].str.replace(',', ', ', regex=False)
```

```python
# limpieza de 'popular_tags'
df['languages'] = df['languages'].str.replace(',', ', ', regex=False)
```

```python
#Limpieza de la columna 'release_date'
# Elimina filas que no contienen un año de 4 dígitos (TBA, TBD, Coming Soon, etc.)
df = df[df['release_date'].str.contains(r'\d{4}', na=False)]

# Maneja casos como '20180108' convirtiéndolos a '2018', luego toma el último año en casos como '2019/2020'
df['release_date'] = df['release_date'].str.replace(r'(\d{4})\d{4}', r'\1', regex=True).str.findall(r'\d{4}').str[-1]

# Extrae el último grupo de 4 dígitos como año final
df['release_date'] = df['release_date'].str.extract(r'(\d{4})(?!.*\d{4})')

# Convierte el año a tipo entero
df['release_date'] = df['release_date'].astype(int)
```

```python
#df['game_description'] = df['game_description'].str.replace('about this game|about this content', '', case=False, regex=True).str.strip()


def clean_text(text):
    if pd.isna(text):
        return text
    text = re.sub(r"\*+|\s+", ' ', text)
    text = text.lower()
    return text.strip()

df['game_description'] = df['game_description'].apply(clean_text)
```

```python
#Limpieza de la columna 'original_price'
# Step 1: Replace 'Free' and '0' with '0'
df['original_price'] = df['original_price'].replace({'Free': '0', '0': '0'})

# Step 2: Remove dollar signs
df['original_price'] = df['original_price'].astype(str).str.replace('$', '', regex=False)

# Step 3: Convert to float, coercing errors to NaN
df['original_price'] = pd.to_numeric(df['original_price'], errors='coerce')

# Optional: fill NaN values with 0.0 if needed
df['original_price'] = df['original_price'].fillna(0.0)
```

```python
# Extrae el ID numérico del juego desde la URL de Steam
df['url']=df['url'].str.extract(r'/app/(\d+)/')
```

```python
df['review_percentage'] = df['review_percentage'].apply(lambda x: f"Porcentaje de recomendación de jugadores: {int(x)}%" if pd.notna(x) else x)
```

```python
df['popular_tags'] = df['popular_tags'].apply(lambda x: f"Tags populares: {x}" if pd.notna(x) else x)
```

```python
df['game_details'] = df['game_details'].apply(lambda x: f"Tags populares: {x}" if pd.notna(x) else x)
```

```python
df['genre'] = df['genre'].apply(lambda x: f"Genero de juego: {x}" if pd.notna(x) else x)
```

```python
df['embedding'] = (
    df['game_description'] + '\n' +
    df['genre'] + '\n' +
    df['popular_tags'] + '\n' +
    df['game_details'] + '\n' +
    df['review_percentage']
)
```

```python
from sentence_transformers import SentenceTransformer, util
import torch

# 1. Configuración del Modelo
# Usamos un modelo balanceado entre velocidad y precisión
model_name = 'all-MiniLM-L6-v2'
model = SentenceTransformer(model_name)

# 2. Simulación de Datos (Basado en el dataset de Steam de Kaggle)
# Cuando el equipo de datos te pase el CSV, cambiaremos esto por pd.read_csv()
data = {
    'game_id': [1, 2, 3, 4, 5],
    'title': ['Left 4 Dead 2', 'Stardew Valley', 'Elden Ring', 'Portal 2', 'Cyberpunk 2077'],
    'description': [
        'Zombies apocalypse cooperative shooter.',
        'Country life RPG with farming and animals.',
        'Challenging action RPG in a dark fantasy world.',
        'Physics-based puzzle game with portals.',
        'Open world action RPG set in a futuristic city.'
    ],
    'tags': ['Action, Zombies, Co-op', 'Farming, Simulation, Relaxing', 'Souls-like, Difficult, RPG', 'Puzzle, Sci-fi, Funny', 'Cyberpunk, RPG, Open World']
}

df = pd.DataFrame(data)

# 3. Creación del "Vibe Text" (Adaptación sugerida)
# SBERT funciona mejor si combinamos la descripción con los tags
df['metadata_combined'] = df['title'] + " " + df['description'] + " " + df['tags']

# 4. Generación de Embeddings
print(f"Vectorizando {len(df)} juegos...")
game_embeddings = model.encode(df['metadata_combined'].tolist(), convert_to_tensor=True)

# 5. Función de Recomendación Adaptada
def get_recommendations(user_query, n_top=3):
    # Vectorizar la consulta del usuario
    query_embedding = model.encode(user_query, convert_to_tensor=True)

    # Calcular similitud coseno
    cosine_scores = util.cos_sim(query_embedding, game_embeddings)[0]

    # Obtener los mejores N resultados
    top_results = torch.topk(cosine_scores, k=n_top)

    print(f"--- Recomendaciones para: '{user_query}' ---\n")
    for score, idx in zip(top_results.values, top_results.indices):
        game = df.iloc[idx.item()]
        print(f"🎯 Juego: {game['title']}")
        print(f"📊 Similitud: {score:.4f}")
        print(f"📝 Tags: {game['tags']}")
        print("-" * 40)

# PRUEBA DE FUNCIONAMIENTO
get_recommendations("I want a difficult game with magic and swords")
```
